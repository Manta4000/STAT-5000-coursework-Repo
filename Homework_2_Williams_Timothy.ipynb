## Problem 1 (20 points) Some probability calculations

**PART A:** The CU Boulder triathlon team has $12$ women and $9$ men. The team is going to a race and can only enter $5$ participants. What is the probability of randomly selecting a race squad of $5$ participants with exactly $3$ women?

```r
tot_women <- 12
tot_men <- 9
full_roster <- tot_women + tot_men

p_first_draw <- tot_women / full_roster

p_sec_draw <- (tot_women - 1) / (full_roster - 1)

p_third_draw <- (tot_women - 2) / (full_roster -2)

p_three_women_drawn <- p_first_draw * p_sec_draw * p_third_draw

print("The odds of three women selected for the race squad, regardless of order should be:")
print(p_three_women_drawn)
print("or approximately:")
print(round(p_three_women_drawn * 100, 2))
print("percent")


```
