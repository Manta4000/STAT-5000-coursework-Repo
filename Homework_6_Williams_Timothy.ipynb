## Problem 1 (10 points) Approximate Confidence Interval for Proportions

Recall from an earlier assignment that if $np > 5$ and $n(1-p) > 5$, then $X \sim \text{Bin}(n,p)$ is well-approximated by $Y \sim N\left(np, np(1-p)\right)$. Use this approximation in the question below. In particular, an approximate $(1-\alpha)\times 100\%$ confidence interval for a population proportion $p$ is given by

\begin{align*}
\widehat{p} \pm z_{1-\alpha/2}\sqrt{\frac{\widehat{p}(1-\widehat{p})}{n}},
\end{align*}

where $\widehat{p}$ is the sample proportion. 

Suppose that 12 people in a sample of 95 are members of the Green Party. Calculate an approximate 90% confidence interval for the true proportion of Green Party members in the population. Interpret this interval.


$\widehat p=\frac{12}{95}\approx 0.1263.$

For a 90% confidence interval, $(z_{1-\alpha/2}=z_{0.95}\approx 1.645)$. Thus,
$\widehat p\pm z_{0.95}\sqrt{\frac{\widehat p(1-\widehat p)}{95}}$
$= 0.1263\pm1.645\sqrt{\frac{0.1263(1-0.1263)}{95}}$
$\approx 0.1263\pm0.0561$.

The approximate 90% confidence interval is

$\boxed{(0.070,\ 0.182)}$.

We are 90% confident that the true population proportion of Green Party members is between approximately $(7.0 \space percent)$ and $18.2 percent$.

## Problem 2 (10 points) Speed of Light

In 1881 Michelson and Newcomb measured the time light took to travel a distance of $7400$ meters. From a study of their experimental setup and a descriptive study of their $64$ measurements, we conclude that the data can be assumed to be i.i.d. These measurements yield the following sample quantities in microseconds for the sample mean $\bar x$ and sample standard deviation $s$:
$$\bar{x}=27.75, s=5.08$$
Construct a 95% confidence interval for the time light takes to travel $7400$ meters.

The population standard deviation is unknown, so use a \(t\)-interval with \(n-1=63\) degrees of freedom. For a 95% confidence interval, $(t_{0.975,63}\approx 1.998)$.

$
\bar{x}\pm t_{0.975,63}\frac{s}{\sqrt{n}}
=27.75\pm 1.998\frac{5.08}{\sqrt{64}}
=27.75\pm 1.27.
$

Thus, the 95% confidence interval is

\[
\boxed{(26.48,\ 29.02)\text{ microseconds}}.
\]

We are 95% confident that the mean time light takes to travel 7400 meters is between approximately \(26.48\) and \(29.02\) microseconds.

## Problem 3 (10 points) A Change in Confidence

A journal article reports that a sample of size $n = 5$ was used as a basis for calculating a $95\%$ CI for the true average natural frequency (Hz) of delaminated beams of a certain type. The resulting interval was $(229.764, 233.504)$. You decide that a confidence level of 99% is more appropriate than the 95% level used.

What are the limits of the 99% interval?


The original 95% confidence interval has midpoint and half-width

$
\bar{x}=\frac{229.764+233.504}{2}=231.634,
\qquad
h_{95}=\frac{233.504-229.764}{2}=1.870.
$

With $n=5$, there are $4$ degrees of freedom. The 95% and 99% critical values are $t_{0.975,4}\approx2.776$ and $t_{0.995,4}\approx4.604$. Keeping the same standard error, the 99% half-width is

$
h_{99}=1.870\frac{4.604}{2.776}\approx3.101.
$

Therefore, the 99% confidence interval is

$
\boxed{(231.634-3.101,\ 231.634+3.101)
\approx(228.533,\ 234.735)\text{ Hz}.}

## Problem 4 (20 points) MLEs

Suppose that $X_1,...,X_n \overset{iid}{\sim}N(\mu, \sigma^2)$, where $\sigma$ is known, and we are ultimately interested in an estimator for $\theta = \mu^2$.

**PART A:** First, find the MLE of $\mu$.

The likelihood is
$
L(\mu)=\prod_{i=1}^n \frac{1}{\sqrt{2\pi\sigma^2}}
\exp\!\left(-\frac{(X_i-\mu)^2}{2\sigma^2}\right).
$
Its log-likelihood, omitting terms that do not depend on $\mu$, is
$
\ell(\mu)=-\frac{1}{2\sigma^2}\sum_{i=1}^n(X_i-\mu)^2.
$
Differentiating and setting the result to zero gives
$
\frac{d\ell}{d\mu}
=\frac{1}{\sigma^2}\sum_{i=1}^n(X_i-\mu)=0
\quad\Longrightarrow\quad
\widehat{\mu}_{\mathrm{MLE}}=\bar X.
$
Since the log-likelihood is maximized at this value, the MLE is
$
\boxed{\widehat{\mu}_{\mathrm{MLE}}=\bar X=\frac{1}{n}\sum_{i=1}^n X_i.}
$

**PART B:** Find the maximum likelihood estimator (MLE) for $\theta$, denoted $\widehat{\theta}$.

This should be easy!

By the invariance property of maximum likelihood estimators, the MLE of $\theta=\mu^2$ is the square of the MLE of $\mu$. Since $\widehat{\mu}_{\mathrm{MLE}}=\bar X$,

$
\boxed{\widehat{\theta}_{\mathrm{MLE}}=\left(\widehat{\mu}_{\mathrm{MLE}}\right)^2=\bar X^2.}
$

**PART C:** Compute the bias of $\widehat{\theta}$, denoted $\text{Bias}(\widehat{\theta})$. Recall that $Bias(\widehat{\theta}) = E(\widehat{\theta}) - \theta$.

Since $\widehat{\theta}=\bar X^2$, and $\bar X$ has mean $\mu$ and variance $\sigma^2/n$,

$
E(\widehat{\theta})
=E(\bar X^2)
=\operatorname{Var}(\bar X)+[E(\bar X)]^2
=\frac{\sigma^2}{n}+\mu^2.
$

Therefore, the bias is

$
\operatorname{Bias}(\widehat{\theta})
=E(\widehat{\theta})-\theta
=\left(\frac{\sigma^2}{n}+\mu^2\right)-\mu^2
=\boxed{\frac{\sigma^2}{n}}.
$

## Problem 5 (25 points) Hubble Data

Load `hubble.csv` into `R`. A description of the variables can be obtained from page 73 of https://cran.r-project.org/web/packages/gamair/gamair.pdf.

**PART A:** Calculate the $85\%$ confidence interval for the mean of a galaxy's distance from Earth in megaparsecs in `R` by doing the computation explicitly.

```r
hubble <- read.csv("hubble.csv")

distance <- na.omit(hubble$distance)
n <- length(distance)
x_bar <- mean(distance)
s <- sd(distance)

alpha <- 0.15
t_crit <- qt(1 - alpha / 2, df = n - 1)
margin_error <- t_crit * s / sqrt(n)
ci_85 <- c(x_bar - margin_error, x_bar + margin_error)

cat("Sample mean:", x_bar, "\n")
cat("85% confidence interval:", ci_85[1], "to", ci_85[2], "megaparsecs\n")
```

**PART B:** Find a built-in `R` function that does this computation automatically and verify that the built in `R` function does the same thing as the confidence interval formula used in part (a).

```r
# Compute the 85% confidence interval using R's built-in t.test() function
ci_85_builtin <- t.test(distance, conf.level = 0.85)$conf.int

cat("85% CI from t.test():", ci_85_builtin[1], "to", ci_85_builtin[2], "megaparsecs\n")
cat("Matches the explicit calculation:", isTRUE(all.equal(
    unname(ci_85_builtin), unname(ci_85)
)), "\n")
```

**PART C:** Interpret the confidence interval.

The 85% confidence interval from Parts A and B gives a range of plausible values for the **population mean distance** of galaxies. We are 85% confident that the true mean distance lies within this interval. In repeated sampling, about 85% of intervals constructed this way would contain the true population mean. This does not mean that 85% of individual galaxies are within the interval.
