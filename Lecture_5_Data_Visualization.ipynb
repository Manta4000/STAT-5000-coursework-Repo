```r
# Putting everything from this exercise together in one cell and ADDING a LEGEND:
ggplot(weather_data, aes(x=c(min(TMIN),max(TMAX))))+
    geom_histogram(aes(x=TMAX, fill="Max Temp"), color="black", alpha=0.2)+
    geom_histogram(aes(x=TMIN, fill="Min Temp"), color="black", alpha=0.2)+
    geom_vline(aes(xintercept=mean(TMAX)), color="blue", linetype = "dashed", linewidth=1)+
    geom_vline(aes(xintercept=mean(TMIN)), color="red", linetype = "dashed", linewidth=1)+
    labs(title="Maximum and Minimum Temperature Distributions in Boulder Co 2017", x="Temp (F)", y="Count")+
    scale_fill_manual(values=c("blue","darkmagenta"),"Data")
```
