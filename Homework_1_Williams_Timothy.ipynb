## Problem 2 (25 points)

Let $x_1, x_2, \ldots, x_n$ be $n$ observations of a variable of interest.  Recall that the sample mean $\bar{x}_n$ and sample variance $s^2_n$ are given by 

$$
\bar{x}_n = \frac{1}{n}\sum_{k=1}^n x_k \quad \textrm{and} \quad s^2_n = \frac{1}{n-1}\sum_{k=1}^n \left( x_k - \bar{x}_n\right)^2 \qquad \tag{Equation 1}
$$

where here the subscript $n$'s indicate the number of observations in the sample. Notice that a natural computation of the variance requires two passes over the data: one to compute the mean, and a second to subtract the mean from each observation and compute the sum of squares. It is often useful to be able to compute the mean in a single pass, inspecting each value $x_k$ only once; for example, when the data are being collected without enough storage to keep all the values, or when costs of memory access dominate those of computation. In this problem you will explore two methods for such an _online_ computation of the mean.  

**PART A**: Show algebraically that the following relation holds between the mean of the first $n-1$ observations and the mean of all $n$ observations: 
$$
\bar{x}_n = \bar{x}_{n-1} + \frac{x_n - \bar{x}_{n-1}}{n}
$$

Note that you can get an expression for $\bar{x}_{n-1}$ by simply replacing $n$ in Equation 1 above with $n-1$.

**PART A:** Show algebraically that

$$\bar{x}_n = \bar{x}_{n-1} + \frac{x_n - \bar{x}_{n-1}}{n}.$$


Starting from

$$\bar{x}_n = \frac{1}{n}\sum_{k=1}^n x_k = \frac{1}{n}\left(\sum_{k=1}^{n-1} x_k + x_n\right),$$


we use the definition of the mean of the first $n-1$ observations:

$$\sum_{k=1}^{n-1} x_k = (n-1)\bar{x}_{n-1}.$$


So

$$\bar{x}_n = \frac{(n-1)\bar{x}_{n-1} + x_n}{n} = \frac{n\bar{x}_{n-1} - \bar{x}_{n-1} + x_n}{n} = \bar{x}_{n-1} + \frac{x_n - \bar{x}_{n-1}}{n}.$$

**PART B:** Implement a function called `update_mean` that implements the formula whose validity you proved in PART A. Note that this function will need to take as its input three things: $x_n$, $\bar{x}_{n-1}$ and $n$. 

Use this function to compute the values that you get from taking the mean of the array given below. Print all 20 estimates to see how the mean is updated with each iteration. Use a built in mean function to compare your final answer using your function with the built-in mean's answer.

$$ x = [5, 8, 9, 23, 15, 17, 68, 27, 43, 21, 33, 12, 30, 31, 33, 42, 41, 40, 7, 5] $$

```r
update_mean <- function(x_n, xbar_prev, n) {
  xbar_prev + (x_n - xbar_prev) / n
}

x <- c(5, 8, 9, 23, 15, 17, 68, 27, 43, 21, 33, 12, 30, 31, 33, 42, 41, 40, 7, 5)

running_mean <- numeric(length(x))
running_mean[1] <- x[1]

for (i in 2:length(x)) {
  running_mean[i] <- update_mean(x[i], running_mean[i - 1], i)
}

running_mean
mean(x)
running_mean[length(running_mean)]
```

The running mean updates with each new observation using the incremental formula. The final value matches the ordinary sample mean, which confirms the online update formula is correct.

Let's create a variable, `v`, that has a bimodal distribution. To do this, we'll make use of the following functions:

- `replicate()`. We used this function in the Introduction to Jupyter and R Notebook from the first week. It's a short way to write loops!
- `ifelse()`. This function will provide different tasks based on an output. For example, `ifelse(a < 5, 1, 0)` will output `1` if `a < 5` and `0` otherwise.
- `runif(n, 0, 1)`. This function generates `n` random numbers between `0` and `1`.
- `rnorm(n,a,b)`. We used this function in the Introduction to Jupyter and R Notebook from the first week. It generates `n` random numbers according to a bell curve, centered over `a`, with "spread" `b`

Ok, here's how we'll create `v`: 

1. Generate a single random number between `0` and `1`. 
2. If the number from the previous step is greater than `0.5`, generate a random number from the bell curve centered at `0` with spread `1`. On the other hand, if the number from the previous step is less than `0.5`, generate a random number from the bell curve centered at `5` with spread `1`.
2. Repeat this process `n = 1000` times.

Then, create a histogram of `v`, and set `bins = 25`. What do you notice about the distribution?

```r
v <- replicate(1000, {
  u <- runif(1)
  if (u > 0.5) {
    rnorm(1, mean = 0, sd = 1)
  } else {
    rnorm(1, mean = 5, sd = 1)
  }
})

hist(v, breaks = 25, col = "purple", main = "Bimodal Distribution of v", xlab = "v")
```

The histogram of $v$ has two peaks, one near 0 and one near 5, which is exactly the bimodal pattern we wanted. This shows that the distribution is not bell-shaped; instead, it is a mixture of two normal distributions centered at different values.
