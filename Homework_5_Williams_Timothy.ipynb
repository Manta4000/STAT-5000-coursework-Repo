## Problem 1 (15 points)

Unfortunately, the expected value of a random variable may not be defined. For example, if $\theta$ is an angle uniformly distributed in the interval $(-\pi/2,\pi/2)$, then
$X=\alpha\tan(\theta)$ has the following distribution:

$$ f(x)=\frac{\alpha}{\pi}\frac{1}{x^2+\alpha^2}, $$

where $\alpha > 0$ is a constant to ensure that $\int_{-\infty}^\infty f(x) \, dx = 1$. Show that $E[X]$ is undefined (HINT: compute $E[X]$ using a $u$-substitution and show that the integral does not converge).

### Completing the antiderivative w/ following steps

$\int_{-\infty}^{\infty} f(x)dx = \int_{-\infty}^{\infty}\frac{\alpha}{\pi}\frac{1}{x^2+\alpha^2}dx$

$= \frac{\alpha}{\pi} \int_{-\infty}^{\infty} \frac{1}{x^2+\alpha^2}dx$ then use $u$-substitution,

$ u = x^2+\alpha^2$

$ du = 2xdx$

$ = \frac{\alpha}{\pi}\int_{-\infty}^{\infty}\frac{1}{u}\frac{1}{2}du$

$ =\frac{1}{2}\frac{\alpha}{\pi}\int_{-\infty}^{\infty}\frac{1}{u}du$

$ =\frac{\alpha}{2\pi}[ln(u)]_{-\infty}^{\infty}du$

$=\frac{\alpha}{2\pi}[\infty - undef.]$ This expression does _not converge_.

## Problem 3 (5 points) Covariance and independence

In class, we introduced covariance and the correlation coefficient to measure how strongly two random variables are related to each other. We know that two dependent random variables can be uncorrelated, but it turns out that if $X$ and $Y$ are independent random variables, then they are uncorrelated: $\text{Corr}[X, Y] = \text{Cov}[X, Y] = 0$.

Suppose that X and Y are independent random variables. Prove that $E[XY] = E[X]E[Y]$.

Since $X$ and $Y$ are independent, their joint distribution factors into the product of their marginal distributions. Thus,

$$
E[XY]
= \int_{\mathbb{R}}\int_{\mathbb{R}} xy\,dF_X(x)\,dF_Y(y)
= \left(\int_{\mathbb{R}} x\,dF_X(x)\right)
    \left(\int_{\mathbb{R}} y\,dF_Y(y)\right)
= E[X]E[Y].
$$

Therefore, $E[XY]=E[X]E[Y]$.

## Problem 4 (20 points) Properties of $\bar{X}$

Let $ X_{1},X_{2},...,X_{n}$ be an independent and identically distributed sample from the distribution $N( \mu , \sigma^{2})$.

**PART A:** Show (without integration), that $sd[\bar{X}] = \frac{\sigma}{\sqrt{n}}$

Because the \(X_i\) are independent and each has variance \(\sigma^2\),

$$
\operatorname{Var}(\bar X)
= \operatorname{Var}\left(\frac{1}{n}\sum_{i=1}^n X_i\right)
= \frac{1}{n^2}\sum_{i=1}^n \operatorname{Var}(X_i)
= \frac{n\sigma^2}{n^2}
= \frac{\sigma^2}{n}.
$$

Therefore,

$$
\operatorname{sd}(\bar X)
= \sqrt{\operatorname{Var}(\bar X)}
= \frac{\sigma}{\sqrt{n}}.
$$

**PART B:** Show that $\displaystyle\bar{X} = \mu + \frac{\sigma}{\sqrt{n}}Z$ where $Z \sim N(0,1)$.

Since the \(X_i\) are independent normal random variables, their average is also normal. From Part A,

$$
\bar X \sim N\left(\mu,\frac{\sigma^2}{n}\right).
$$

Standardize \(\bar X\) by defining

$$
Z=\frac{\bar X-\mu}{\sigma/\sqrt{n}}.
$$

Then \(Z\sim N(0,1)\). Rearranging gives

$$
\boxed{\bar X=\mu+\frac{\sigma}{\sqrt{n}}Z,\qquad Z\sim N(0,1).}
$$

**PART C:** (OPTIONAL for STAT 4000) Use PART B to find $\displaystyle Var[\bar{X}^{2}]$. Assume that $Cov[Z, Z^2] = 0$, and that $\text{Var}[Z^2]  = 2$ (you don't have to prove these facts).

Using Part B, let \(a=\sigma/\sqrt{n}\). Then

$$
\bar X^2=(\mu+aZ)^2
=\mu^2+2\mu aZ+a^2Z^2.
$$

The constant does not affect variance. Using \(\operatorname{Var}(Z)=1\), \(\operatorname{Var}(Z^2)=2\), and \(\operatorname{Cov}(Z,Z^2)=0\),

$$
\begin{aligned}
\operatorname{Var}(\bar X^2)
&=\operatorname{Var}(2\mu aZ+a^2Z^2)\\
&=4\mu^2a^2\operatorname{Var}(Z)
    +a^4\operatorname{Var}(Z^2)
    +4\mu a^3\operatorname{Cov}(Z,Z^2)\\
&=4\mu^2a^2+2a^4\\
&=\boxed{\frac{4\mu^2\sigma^2}{n}+\frac{2\sigma^4}{n^2}}.
\end{aligned}
$$

## Problem 6 (15 points)

Let $X_{1},X_{2},...,X_{500}$ be an independent and identically distributed sample from the distribution $N( \mu = 5 , \sigma^{2} = 4)$.

Construct a matrix `x` with $m = 10,000$ columns, where each column is a different sample (i.e., set of realizations) from $ X_{1},X_{2},...,X_{500}$. Then, use `colMeans()` to find the mean of each sample/column. Store these means in `xbar`. Based on lecture, what is the distribution of `xbar`? Is a histogram of `xbar` consistent with that?

```r
set.seed(123)

x <- matrix(
    rnorm(500 * 10000, mean = 5, sd = 2),
    nrow = 500,
    ncol = 10000
)

xbar <- colMeans(x)

hist(
    xbar,
    probability = TRUE,
    main = "Distribution of Sample Means",
    xlab = "Sample mean"
)
curve(
    dnorm(x, mean = 5, sd = sqrt(4 / 500)),
    add = TRUE,
    col = "red",
    lwd = 2
)
```
