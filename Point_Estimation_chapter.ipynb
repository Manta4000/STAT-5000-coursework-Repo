## Properties of Point Estimators

Statistics of functions of Random variables and are therefore also random variables

## Mean Square Error

The difference between an estimator $T$ for an unknown parameter $\theta$ and the parameter itself is the error.

This quantity can be negative, thus the error is often sqared to conduct comparisons.

$MSE[T] = E[(T-\theta)^2]$

The MSE consists of two non-negative components; the variance estimator and the squared bias estimator.

The variance estimator:

$Var[T] = E[(T-E[T])^2]$

The Squared-bias estimator:

$E[T] - \theta$
