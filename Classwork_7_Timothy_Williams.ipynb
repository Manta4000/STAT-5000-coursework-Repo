$\frac{1}{n-1}E[\sum^{n}_{i=1}(x_i-\bar{x})^2]$ in which the sum should be the same as the expected value of $\sigma^2$

## Problem #5

In this example, we will construct a simulation to verify the “coverage properties” of a confidence interval for the mean of a normal distribution.

**5.(a) Simulate a matrix with `m = 1000` rows and `n = 100` columns, where each entry is a random number from the population model $N(0, 1)$. Interpret each row as a sample from the population.**


```r
# 9) CODE HERE
m <- 1000
n <- 100
x <- matrix(rnorm(m * n, mean = 0, sd = 1), nrow = m, ncol = n)
```

**5.(b) Suppose that we didn’t know the population mean, $\mu$, but wanted to estimate it using a confidence interval. For each sample, calculate the $95\%$ confidence for the mean, $\mu$. Assume $\sigma = 1$ is known.**

```r
# 10) CODE HERE
sample_means <- rowMeans(x)
margin_of_error <- qnorm(0.975) * 1 / sqrt(n)

confidence_intervals <- cbind(
    lower = sample_means - margin_of_error,
    upper = sample_means + margin_of_error
)
```

**5.(c) Why would we use a confidence interval instead of just reporting the sample mean $\bar{x}$?**

11) Answer Here
A sample mean $\bar{x}$ is a point estimate, but it does not show how much uncertainty there is due to sampling variability. A confidence interval provides a range of plausible values for the population mean and indicates the estimate’s precision. For example, a 95% confidence interval is produced by a method that captures the true mean in about 95% of repeated samples.

**5.(d) Print and interpret the confidence interval for the first sample (i.e., when `m = 1`).**

```r
# 12) CODE HERE
first_interval <- confidence_intervals[1, ]
print(first_interval)

cat(
    "The 95% confidence interval for the first sample is (",
    round(first_interval["lower"], 3), ", ",
    round(first_interval["upper"], 3), ").\n",
    "This interval estimates the true mean μ using the sample mean ",
    round(sample_means[1], 3), " and a margin of error of ",
    round(qnorm(0.975) * 1 / sqrt(n), 3), ".\n"
)
```
