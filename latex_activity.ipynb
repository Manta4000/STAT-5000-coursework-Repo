For this class, you can always turn in scans of handwritten work. However, it might be worth your time to learn some Latex commands.

- Create a new notebook (either locally or using an online resource like the CSEL site)
- Create three Markdown cells
- Use each cell to replicate the three cells below *exactly*.

I recommend making an account on [overleaf.com](www.overleaf.com) so that you can edit `.tex` files. Also, take a look at the files `latex_cheat_sheet.pdf` and `latex_examples.tex` on Canvas for examples on how to make certain symbols.

Cell #1:
Sometimes, it makes sense to put an expression in the same line as text, like this: $a^2 + b^2 - 2ab \cos \theta = c^2$.

Cell #2: Other times, it makes sense to set an expression apart from the text, like this:
$$\int_a^b f(x) \, dx = F(b) - F(a)$$

Cell #3: Another useful thing is to align equations, which makes your work easier to follow:
\begin{align}
    \frac{1}{2} \sum_{i = 1}^n (a_i + b_i) &= \frac{1}{2} \left( \sum_{i = 1}^n a_i + \sum_{i = 1}^n b_i \right) \\
    &= \frac{1}{2} \sum_{i = 1}^n a_i + \frac{1}{2} \sum_{i = 1}^n b_i.
\end{align}
