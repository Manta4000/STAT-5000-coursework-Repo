## Problem 5 (25 points) Mixture distributions

Most of the distributions we've looked at so far have been unimodal, flat, or monotonically increasing/decreasing. Let's take a look at a class of distributions that are multimodal.

Let $X_1,...,X_n$ be continuous random variables with PDFs $f_{X_i}(x)$, $i = 1,...,n$. Define a *mixture distribution* to be a random variable $X$ with pdf

\begin{align*}
f_X(x) = \sum^n_{i=1}\alpha_i f_{X_i}(x) 
\end{align*}

where $\alpha_i$ are nonnegative real numbers such that $\sum^n_{i=1}\alpha_i = 1$. You can think of $f_X(x)$ as being a weighted average of the PDFs $f_{X_i}(x)$.

**PART A:** Show that $f_X(x)$ is a PDF.

Since each $\alpha_i \ge 0$ and each $f_{X_i}(x)$ is a valid pdf, we have $f_{X_i}(x) \ge 0$ for all $x$. Therefore

$$
f_X(x)=\sum_{i=1}^n \alpha_i f_{X_i}(x) \ge 0.
$$

Also, because the pdfs integrate to 1 and the weights sum to 1,

$$
\int_{-\infty}^{\infty} f_X(x)\,dx
= \int_{-\infty}^{\infty} \sum_{i=1}^n \alpha_i f_{X_i}(x)\,dx
= \sum_{i=1}^n \alpha_i \int_{-\infty}^{\infty} f_{X_i}(x)\,dx
= \sum_{i=1}^n \alpha_i(1)
= 1.
$$

Thus $f_X(x)$ is nonnegative and integrates to 1, so it is a valid probability density function.

**PART B:** Find the expected value of $X$.

For a mixture distribution,
$$
f_X(x)=\sum_{i=1}^n \alpha_i f_{X_i}(x),
$$
the expected value is
$$
E[X]=\int_{-\infty}^{\infty} x f_X(x)\,dx
= \int_{-\infty}^{\infty} x \sum_{i=1}^n \alpha_i f_{X_i}(x)\,dx
= \sum_{i=1}^n \alpha_i \int_{-\infty}^{\infty} x f_{X_i}(x)\,dx.
$$

Since
$$
\int_{-\infty}^{\infty} x f_{X_i}(x)\,dx = E[X_i] = \mu_i,
$$
we get
$$
E[X] = \sum_{i=1}^n \alpha_i \mu_i.
$$

So the mean of a mixture is a weighted average of the component means.

````r
### A minimal Monte Carlo solution is

f <- function(x) {
    ifelse(0 <= x & x <= pi, (cos(x) + 1) / pi, 0)
}

integrate <- function(num_samples, ub, isPlot = FALSE) {
    x <- runif(num_samples, 0, ub)
    y <- runif(num_samples, 0, 1)
    z <- as.integer(y < f(x))
    
    est <- ub * mean(z)
    var_z <- var(z)
    
    if (isPlot) {
        xs <- seq(0, ub, length.out = 200)
        plot(xs, f(xs), type = "l", lwd = 2, col = "blue",
                 xlab = "x", ylab = "f(x)", main = "Monte Carlo integration")
        points(x, y, col = ifelse(z == 1, "green", "red"),
                     pch = 16, cex = 0.6)
    }
    
    list(estimate = est, sample_variance_Z = var_z)
}

### Here Z = I(Y < f(X)) is Bernoulli, so the estimator is ub * mean(Z), which estimates the integrand.
````
